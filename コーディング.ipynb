```python
def f(lst):
    result = []
    for i in range(0,len(lst)-2):
        for j in range(1,len(lst)-2):
            if (lst[i] != lst[j]):
                result.append(lst[j])
    return result

f(
     ['a','b','c','a','c','d'] # 6
)

# [0] [1]
# [0] [2]
# [0] [3]
# .......
# [0] [i]

# [1] [1]
# [1] [2]
# [1] [3]
# .......
# [1] [i]
# .......
# .......
# [j] [i]

# max(j) -> [0,4] -> range(0,len(lst)-2)
# max(i) -> [1,5] -> range(1,len(lst)-1)
```

```python
def f(lst):
    result = []

    for i in range(len(lst)): # 0,1,2,3,..,i,...len(lst)-1
        for j in range(0,i): #0,1,2,3,..i-1
            if (lst[j] == lst[i]):
                break
        else:
            result.append(lst[i])
    return result



f(
    ['a','b','c','a','c','d'] # 6
)


```
